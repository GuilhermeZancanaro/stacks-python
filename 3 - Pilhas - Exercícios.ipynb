1. Implemente uma classe chamada Pilha que simule o comportamento de uma pilha (LIFO - Last In, First Out). A classe deve conter:
   * Um método push(elemento) que adiciona um elemento ao topo da pilha.
   * Um método pop() que remove e retorna o elemento do topo da pilha.
   * Um método top() que retorna o elemento do topo da pilha sem removê-lo.
   * Um método is_empty() que verifica se a pilha está vazia.
   * Um método size() que retorna o tamanho da pilha.

* Crie uma pilha, adicione alguns elementos, remova o topo e exiba o tamanho da pilha.

```python
# resolva o exercício aqui
class Pilha:
    def __init__(self):
        self.itens = []

    def push(self,elemento):
        self.itens.append(elemento)

    def pop(self):
        if not self.is_empty():
            return self.itens.pop()
        else:
            return None

    def top(self):
        if not self.is_empty():
            return self.itens[-1]
        else:
            return None

    def is_empty(self):
        return len(self.itens) == 0

    def size(self):
        return len(self.itens)

pilha = Pilha()

pilha.push(1)
pilha.push(2)
pilha.push(3)


print("Elemento removido do topo: ", pilha.pop())
print("Lista: ", pilha.size())
```

2. Escreva uma função chamada verificar_parenteses que receba uma string contendo parênteses, colchetes e chaves ((), [], {}) e verifique se a sequência está corretamente balanceada.
Regras:
   * Utilize uma pilha para armazenar os símbolos de abertura.
   * Sempre que encontrar um símbolo de fechamento, verifique se o topo da pilha contém o símbolo de abertura correspondente.
   * No final, a pilha deve estar vazia para que a sequência seja considerada balanceada.
  
* Exemplo:

`Entrada: "{[()]}"`
`Saída esperada: "Sequência balanceada"`

`Entrada: "{[(])}"`
`Saída esperada: "Sequência não balanceada"`

```python
# resolva o exercício aqui

def verificar_parenteses(seq):
    correspondentes = { ')': '(', ']': '[', '}' : '{'}

    pilha = Pilha()

    for caracter in seq:
        if caracter in correspondentes.values():
            pilha.push(caracter)

        elif caracter in correspondentes.keys():
            if pilha.is_empty() or pilha.pop() != correspondentes[caracter]:
                return "sequencia não balanceada"

    return "sequencia balanceada" if pilha.is_empty() else "sequencia não balanceada"


print(verificar_parenteses("{asd[asd(asd)asd]asd}"))
print(verificar_parenteses("{asdasd[dasdh]()"))

```

3. Implemente uma classe chamada EditorDeTexto que simule as operações de um editor de texto simples, com a funcionalidade de desfazer e refazer operações. A classe deve conter:
    1.	Um método escrever(texto) que adiciona texto ao conteúdo atual.
	2.	Um método desfazer() que desfaz a última operação de escrita.
	3.	Um método refazer() que refaz a última operação desfeita (se houver).
	4.	Um método exibir_conteudo() que exibe o conteúdo atual do texto.

```python
# resolva o exercício aqui

class editorDeTexto:

    def __init__(self):
        self.conteudo = ""
        self.historico = Pilha()
        self.refazer_pilha = Pilha()

    def escrever(self, texto):
        self.historico.push(self.conteudo)
        self.conteudo += texto
        self.refazer_pilha = Pilha()

    def desfazer(self):
        if not self.historico.is_empty():

            self.refazer_pilha.push(self.conteudo)
            self.conteudo = self.historico.pop()

    def refazer(self):
        if not self.refazer_pilha.is_empty():
            self.historico.push(self.conteudo)
            self.conteudo = self.refazer_pilha.pop()

    def exibir_conteudo(self):
        return self.conteudo


editor = editorDeTexto()

editor.escrever("Ola ")
editor.escrever("mundo ")

print("Conteudo inteiro: ", editor.exibir_conteudo())

editor.desfazer()
print("Conteudo desfeito de maneira LIFO: ", editor.exibir_conteudo())

editor.refazer()
print("Conteudo refeito recolocando oq foi desfeito: ", editor.exibir_conteudo())
```
